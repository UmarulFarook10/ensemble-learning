```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
```

```python
df=pd.read_csv("C:/Users/space/Desktop/datas/Clustering_gmm (2).csv")
```

```python
df.columns
```

```python
df
```

```python
x=df[['Weight', 'Height']]
```

```python
scalar=StandardScaler()
x_scalar=scalar.fit_transform(x)
print(x_scalar)
```

```python
model=DBSCAN(eps=0.5,min_samples=8)
```

```python

ypred=model.fit_predict(x_scalar)
print(ypred)
```

```python
score=silhouette_score(x,ypred)
print(score)
```

```python
label=model.labels_
print(label)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.Figure(figsize=(8,6))
plt.scatter(x['Weight'],x['Height'],c=label,cmap="viridis")
plt.title("dbscan: weight vs 'Height'")
plt.xlabel("weight")
plt.ylabel("'Height'")
plt.show()
```

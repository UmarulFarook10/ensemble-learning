```python
import pandas as pd
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans
```

```python
df=pd.read_csv("C:/Users/space/Desktop/datas/Clustering_gmm (2).csv")
```

```python
df.columns
```

```python
x=df[['Weight', 'Height']]
```

```python
scalar=StandardScaler()
x_scalar=scalar.fit_transform(x)
print(x_scalar)
```

```python
model1=DBSCAN(eps=0.5,min_samples=5)
model2=KMeans(n_clusters=4,max_iter=3)
ypred1=model1.fit_predict(x_scalar)
ypred2=model2.fit_predict(x_scalar)
print(ypred1)
print(ypred2)
```

```python
center=model2.cluster_centers_
print(center)
```

```python
score1=silhouette_score(x,ypred1)
score2=silhouette_score(x,ypred2)
print("DBSCAN score=",score1)
print("KMeans score=",score2)
```

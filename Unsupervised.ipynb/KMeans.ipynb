```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("C:/Users/space/Desktop/datas/Mall_Customers.csv")
```

```python
x=df[["Age","Annual Income (k$)"]]
scalar=StandardScaler()
x_scalar=scalar.fit_transform(x)#standard scalar is used to reduce the complex of data which will easy for computer to evauate
print(x_scalar)
```

```python
model=KMeans(n_clusters=4,max_iter=3)
```

```python
print(x)
```

```python
model.fit(x_scalar)
```

```python
ypred=model.predict(x_scalar)
print(ypred)
```

```python
sil=silhouette_score(x,ypred)
print(sil)
```

```python
label=model.labels_
print(label)
```

```python
centre=model.cluster_centers_
print(centre)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(x['Age'],x['Annual Income (k$)'],c=label,cmap='rainbow',edgecolors='k')
plt.scatter(centre[:,0],centre[:,1],s=300,c='green',marker='*',label='Centroids')
plt.title("K-Means: Age vs Annual Income")
plt.xlabel("Age")
plt.ylabel("Annual Income (k$)")
plt.legend()
plt.grid(True)
plt.show()
```
